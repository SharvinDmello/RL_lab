Using dynamic programming, implement the policy evaluation algorithm to compute the value function for a given policy in a small grid-world environment.

```python
import numpy as np

GRID_SIZE = 4
GAMMA = 0.9
THETA = 0.001

GOAL = (3, 3)
UP = (-1, 0)
DOWN = (1, 0)
LEFT = (0, -1)
RIGHT = (0, 1)
```

```python
def policy(state):
    row, col = state
    if col == GRID_SIZE - 1:
        return DOWN
    return RIGHT

def step(state, action):
    row, col = state
    dr, dc = action

    new_row = row + dr
    new_col = col + dc

    if new_row < 0 or new_row >= GRID_SIZE or new_col < 0 or new_col >= GRID_SIZE:
        return state, -1

    next_state = (new_row, new_col)

    if next_state == GOAL:
        return next_state, 10

    return next_state, -1
```

```python
def policy_evaluation():
    V = np.zeros((GRID_SIZE, GRID_SIZE))
    iteration = 0

    while True:
        delta = 0
        new_V = V.copy()

        for row in range(GRID_SIZE):
            for col in range(GRID_SIZE):
                state = (row, col)

                if state == GOAL:
                    continue

                action = policy(state)
                next_state, reward = step(state, action)
                nr, nc = next_state
                new_value = reward + GAMMA * V[nr, nc]

                delta = max(delta, abs(new_value - V[row, col]))
                new_V[row, col] = new_value

        V = new_V
        iteration += 1

        print(f"\nIteration {iteration}")
        print(np.round(V, 3))

        if delta < THETA:
            break

    return V
```

```python
V = policy_evaluation()

print("\nFinal Value Function:")
print(np.round(V, 3))
```

```python
states = ["A", "B", "C"]
V = {state: 0.0 for state in states}

alpha = 0.1
gamma = 0.9
episodes = 1000
```

```python
for episode in range(episodes):
    state = "A"

    while state != "Terminal":
        if state == "A":
            next_state = "B"
            reward = 0
        elif state == "B":
            next_state = "C"
            reward = 0
        elif state == "C":
            next_state = "Terminal"
            reward = 1

        if next_state == "Terminal":
            V[state] += alpha * (reward - V[state])
        else:
            V[state] += alpha * (reward + gamma * V[next_state] - V[state])

        state = next_state
```

```python
print("Learned State Values:")
for state in states:
    print(f"V({state}) = {V[state]:.4f}")
```

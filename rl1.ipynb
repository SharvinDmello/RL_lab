Implement a basic Markov Decision Process (MDP) using Python libraries such as NumPy and Pandas. Demonstrate the concepts of states, actions, transitions, rewards, and discount factors. Provide a detailed explanation of your implementation.

```python
import numpy as np
import pandas as pd

states = np.array([0, 1, 2, 3, 4])
state_names = {
    0: "Start",
    1: "A",
    2: "B",
    3: "C",
    4: "Goal"
}
num_states = len(states)

actions = np.array([0, 1])
action_names = {
    0: "Left",
    1: "Right"
}
num_actions = len(actions)
```

```python
transition_probabilities = np.zeros((num_states, num_actions, num_states))

transition_probabilities[0, 0, 0] = 1.0
transition_probabilities[0, 1, 1] = 1.0
transition_probabilities[1, 0, 0] = 1.0
transition_probabilities[1, 1, 2] = 1.0
transition_probabilities[2, 0, 1] = 1.0
transition_probabilities[2, 1, 3] = 1.0
transition_probabilities[3, 0, 2] = 1.0
transition_probabilities[3, 1, 4] = 1.0
transition_probabilities[4, 0, 4] = 1.0
transition_probabilities[4, 1, 4] = 1.0

rewards = np.zeros((num_states, num_actions))
rewards[:, :] = -1
rewards[3, 1] = 10
```

```python
print("STATES")
for state in states:
    print(state, "=", state_names[state])

print("\nACTIONS")
for action in actions:
    print(action, "=", action_names[action])

print("\nTRANSITION PROBABILITIES")
for state in states:
    for action in actions:
        for next_state in states:
            probability = transition_probabilities[state, action, next_state]
            if probability > 0:
                print(f"State {state} ({state_names[state]}) + Action {action_names[action]} -> State {next_state} ({state_names[next_state]}) Probability = {probability}")
```

```python
print("\nREWARD TABLE")
reward_table = pd.DataFrame(
    rewards,
    index=[state_names[s] for s in states],
    columns=[action_names[a] for a in actions]
)
print(reward_table)
```

```python
gamma = 0.9
print("\nDISCOUNT FACTOR")
print("Gamma =", gamma)
```

```python
def get_next_state(state, action):
    probabilities = transition_probabilities[state, action]
    return np.argmax(probabilities)

print("\nMDP EPISODE")
current_state = 0
total_return = 0
discount = 1.0
step = 0

while current_state != 4 and step < 10:
    action = 1
    next_state = get_next_state(current_state, action)
    reward = rewards[current_state, action]
    discounted_reward = discount * reward
    total_return += discounted_reward

    print(f"\nStep {step + 1}")
    print("Current State:", current_state, "(", state_names[current_state], ")")
    print("Action:", action_names[action])
    print("Next State:", next_state, "(", state_names[next_state], ")")
    print("Reward:", reward)
    print("Discounted Reward:", round(discounted_reward, 4))

    discount *= gamma
    current_state = next_state
    step += 1

print("\nTotal Discounted Return:", round(total_return, 4))
```

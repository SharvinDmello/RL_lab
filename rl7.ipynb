Implement Every Visit Monte Carlo prediction algorithm using Python and apply it to a simple environment. Display the learned policy and final Q-values.

```python
import numpy as np
import random

n_states = 16
n_actions = 4
gamma = 0.9
episodes = 5000

action_symbols = ['U', 'D', 'L', 'R']

Q = np.zeros((n_states, n_actions))
counts = np.zeros((n_states, n_actions))
```

```python
def choose_action(state, epsilon=0.1):
    if random.random() < epsilon:
        return random.randint(0, 3)

    return np.argmax(Q[state])

def step(state, action):
    row, col = divmod(state, 4)

    if action == 0:
        row = max(0, row - 1)
    elif action == 1:
        row = min(3, row + 1)
    elif action == 2:
        col = max(0, col - 1)
    elif action == 3:
        col = min(3, col + 1)

    next_state = row * 4 + col

    if next_state == 15:
        reward = 10
        done = True
    else:
        reward = -1
        done = False

    return next_state, reward, done
```

```python
for episode in range(episodes):
    state = 0
    trajectory = []
    done = False
    steps = 0

    while not done and steps < 100:
        action = choose_action(state)
        next_state, reward, done = step(state, action)

        trajectory.append((state, action, reward))
        state = next_state
        steps += 1

    G = 0

    for t in reversed(range(len(trajectory))):
        state, action, reward = trajectory[t]
        G = gamma * G + reward

        counts[state, action] += 1
        Q[state, action] += (G - Q[state, action]) / counts[state, action]
```

```python
print("Learned Policy (Every-Visit MC):")
for state in range(n_states):
    if state == 15:
        print(" G ", end=" ")
    else:
        best_action = np.argmax(Q[state])
        print(f" {action_symbols[best_action]} ", end=" ")

    if state % 4 == 3:
        print()

print("\nFinal Q-values:")
print(np.round(Q, 2))
```

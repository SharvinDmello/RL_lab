Define a simple grid world environment where an agent can move in four directions (up, down, left, right). Use NumPy to represent the states, actions, and transition probabilities. Implement a reward system where reaching certain states provides a positive reward, while others provide negative rewards. Demonstrate the concept of the discount factor by calculating the expected return from different starting states.

```python
import numpy as np

grid_size = 3
states = np.arange(grid_size * grid_size)
actions = np.array([0, 1, 2, 3])
num_states = len(states)
num_actions = len(actions)
```

```python
def get_next_state(state, action):
    row = state // grid_size
    col = state % grid_size

    if action == 0:
        row = max(row - 1, 0)
    elif action == 1:
        row = min(row + 1, grid_size - 1)
    elif action == 2:
        col = max(col - 1, 0)
    elif action == 3:
        col = min(col + 1, grid_size - 1)

    next_state = row * grid_size + col
    return next_state
```

```python
transition_probabilities = np.zeros((num_states, num_actions, num_states))
for state in states:
    for action in actions:
        next_state = get_next_state(state, action)
        transition_probabilities[state, action, next_state] = 1.0

goal_state = 8
negative_state = 4
```

```python
def get_reward(next_state):
    if next_state == goal_state:
        return 10
    elif next_state == negative_state:
        return -10
    else:
        return -1

path = [0, 3, 6, 7, 8]
rewards = []

for i in range(len(path) - 1):
    next_state = path[i + 1]
    reward = get_reward(next_state)
    rewards.append(reward)
```

```python
print("GRID WORLD")
print("\nGrid:")
print("0  1  2")
print("3  4  5")
print("6  7  8")
print("\nGoal State:", goal_state)
print("Negative State:", negative_state)
print("\nPath:")
print(" -> ".join(map(str, path)))
print("\nRewards received:")
print(rewards)

return_without_discount = sum(rewards)
print("\nWITHOUT DISCOUNT FACTOR")
print("Return =", return_without_discount)
```

```python
def calculate_discounted_return(rewards, gamma):
    G = 0
    for t, reward in enumerate(rewards):
        discounted_reward = (gamma ** t) * reward
        G += discounted_reward
    return G

gamma_values = [0.5, 0.9, 0.99]
print("\nWITH DISCOUNT FACTOR")

for gamma in gamma_values:
    G = calculate_discounted_return(rewards, gamma)
    print("\nGamma =", gamma)
    for t, reward in enumerate(rewards):
        discounted_reward = (gamma ** t) * reward
        print(f"Step {t + 1}: {reward} * {gamma}^{t} = {discounted_reward:.4f}")
    print("Discounted Return =", round(G, 4))
```

```python
print("\nCOMPARISON")
print(f"{'Method':<25} {'Return':>10}")
print("-" * 35)
print(f"{'Without discount':<25} {return_without_discount:>10.4f}")
for gamma in gamma_values:
    G = calculate_discounted_return(rewards, gamma)
    print(f"{'Gamma = ' + str(gamma):<25} {G:>10.4f}")

print("\nRETURN FROM DIFFERENT STARTING STATES")
paths = {
    7: [7, 8],
    6: [6, 7, 8],
    3: [3, 6, 7, 8],
    0: [0, 3, 6, 7, 8]
}

gamma = 0.9
print("\nUsing gamma =", gamma)
print(f"\n{'Start':<10} {'Path':<25} {'Rewards':<20} {'Return':>10}")
print("-" * 70)

for start_state, path in paths.items():
    path_rewards = []
    for i in range(len(path) - 1):
        next_state = path[i + 1]
        reward = get_reward(next_state)
        path_rewards.append(reward)

    G = calculate_discounted_return(path_rewards, gamma)
    path_string = " -> ".join(map(str, path))
    print(f"{start_state:<10} {path_string:<25} {str(path_rewards):<20} {G:>10.4f}")
```
